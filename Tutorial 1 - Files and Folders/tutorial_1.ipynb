## 3. Getting File Attributes

```python
import os
with os.scandir('my_directory/') as dir_contents:
    for entry in dir_contents:
         info = entry.stat()
         print(info.st_mtime)

```

```python
from pathlib import Path
current_dir = Path('my_directory')
for path in current_dir.iterdir():
    info = path.stat()
    print(info.st_mtime)
    
```

```python
from datetime import datetime
from os import scandir

def convert_date(timestamp):
    d = datetime.utcfromtimestamp(timestamp)
    formated_date = d.strftime('%d %b %Y')
    return formated_date

def get_files():
    dir_entries = scandir('my_directory/')
    for entry in dir_entries:
        if entry.is_file():
            info = entry.stat()
            print(f'{entry.name}\t Last Modified: {convert_date(info.st_mtime)}')
```

```python
get_files()
```

## 6. Traversing Directories and Processing Files

```python
# Walking a directory tree and printing the names of the directories and files
for dirpath, dirnames, files in os.walk('.'):
    print(f'Found directory: {dirpath}')
    for file_name in files:
        print(file_name)
```

```python
for dirpath, dirnames, files in os.walk('.', topdown=False):
    print(f'Found directory: {dirpath}')
    for file_name in files:
        print(file_name)
```

## 7. Making Temporary Files and Directories

```python
from tempfile import TemporaryFile

# Create a temporary file and write some data to it
fp = TemporaryFile('w+t')
fp.write('Hello universe!')

# Go back to the beginning and read data from file
fp.seek(0)
data = fp.read()

# Close the file, after which it will be removed
fp.close()
```

```python
with TemporaryFile('w+t') as fp:
    fp.write('Hello universe!')
    fp.seek(0)
    fp.read()
# File is now closed and removed
```

```python
import tempfile
with tempfile.TemporaryDirectory() as tmpdir:
    print('Created temporary directory ', tmpdir)
    os.path.exists(tmpdir)




# Directory contents have been removed

tmpdir

os.path.exists(tmpdir)
```

## Archiving

### Reading ZIP Files
I learned that the `zipfile` module can open ZIP files and list their contents.

### Extracting ZIP Archives
I learned that `.extract()` extracts one file, and `.extractall()` extracts everything in a ZIP archive.

### Password-Protected Archives
I learned that a password can be passed to `.extract()` or `.extractall()` when opening a protected ZIP file.

### Creating ZIP Archives
I learned that write mode creates a new ZIP archive and append mode adds files without deleting its existing contents.

### Opening TAR Archives
I learned that the `tarfile` module can open TAR archives and different modes are used depending on whether the archive is compressed.

### Extracting TAR Archives
I learned that TAR files can also be extracted individually or all at once using either `.extract()` or `.extractall()`.

### Creating TAR Archives
I learned that files can be added to a new TAR archive with `.add()` and append mode adds files to an existing archive.

### Working With Compressed Archives
I learned that TAR archives can use compression formats like gzip, bzip2, and lzma.

### An Easier Way to Create Archives
I learned that `shutil.make_archive()` and `shutil.unpack_archive()` provide a simpler way to create and extract ZIP or TAR archives.
